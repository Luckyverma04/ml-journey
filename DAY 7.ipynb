```python
import numpy as np
import pandas as pd
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, cross_val_score

np.random.seed(0)
n = 200
free  = np.random.randint(0, 8, n)
links = np.random.randint(0, 6, n)
score = free*1.0 + links*0.8 + np.random.randn(n)*2.5
y = (score > 6).astype(int)
X = np.column_stack([free, links])

for k in [1, 5, 25]:
    knn = KNeighborsClassifier(n_neighbors=k)
    s = cross_val_score(knn, X, y, cv=5, scoring="f1")
    print("K =", k, np.round(s, 2), "average:", round(s.mean(), 2))
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_score, recall_score

np.random.seed(7)
n = 300
free     = np.random.randint(0, 8, n)
links    = np.random.randint(0, 6, n)
caps     = np.random.randint(0, 10, n)
exclaim  = np.random.randint(0, 6, n)
score = 1.0*free + 0.8*links + 0.5*caps + 0.6*exclaim + np.random.randn(n)*3
y = (score > np.percentile(score, 65)).astype(int)
X = np.column_stack([free, links, caps, exclaim])

proba = cross_val_predict(LogisticRegression(), X, y, cv=5, method="predict_proba")[:, 1]

for t in [0.3, 0.5, 0.7]:
    pred = (proba > t).astype(int)
    print("threshold", t,
          "| spam bola:", pred.sum(),
          "| precision:", round(precision_score(y, pred), 2),
          "| recall:", round(recall_score(y, pred), 2))
```
