```python
import numpy as np
import pandas as pd

np.random.seed(7)
n = 300
free     = np.random.randint(0, 8, n)
links    = np.random.randint(0, 6, n)
caps     = np.random.randint(0, 10, n)       # kitne words CAPITAL mein
exclaim  = np.random.randint(0, 6, n)        # kitne '!' 
score = 1.0*free + 0.8*links + 0.5*caps + 0.6*exclaim + np.random.randn(n)*3
y = (score > np.percentile(score, 65)).astype(int)   # ~35% spam
X = np.column_stack([free, links, caps, exclaim])

df = pd.DataFrame(X, columns=["free", "links", "caps", "exclaim"])
df["spam"] = y
print(df.shape)
print(df["spam"].value_counts())
print(df.head())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

models = {
    "Logistic":   LogisticRegression(),
    "Tree d=3":   DecisionTreeClassifier(max_depth=3, random_state=0),
    "Forest":     RandomForestClassifier(n_estimators=100, random_state=0),
    "KNN k=5":    KNeighborsClassifier(n_neighbors=5),
}

for name, m in models.items():
    s = cross_val_score(m, X, y, cv=5, scoring="f1")
    print(name, np.round(s, 2), "avg:", round(s.mean(), 2))
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix, precision_score, recall_score

for name in ["Logistic", "Forest"]:
    pred = cross_val_predict(models[name], X, y, cv=5)
    print(name)
    print(confusion_matrix(y, pred))
    print("precision:", round(precision_score(y, pred), 2),
          "recall:", round(recall_score(y, pred), 2))
    print()
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_score, recall_score

proba = cross_val_predict(models["Logistic"], X, y, cv=5, method="predict_proba")[:, 1]

for t in [0.3, 0.5, 0.7]:
    pred = (proba > t).astype(int)
    print("threshold", t,
          "| spam bola:", pred.sum(),
          "| precision:", round(precision_score(y, pred), 2),
          "| recall:", round(recall_score(y, pred), 2))
```
